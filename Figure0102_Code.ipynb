```python
%matplotlib inline
import numpy as np
from pylab import *
#from sklearn import preprocessing
import seaborn as sns
import pandas as pd
from scipy import signal
from scipy import stats
import pandas as pd
import seaborn as sns
import matplotlib
from matplotlib.lines import Line2D
import os
plt.rcParams['xtick.direction'] = 'in'#将x周的刻度线方向设置向内
plt.rcParams['ytick.direction'] = 'in'#将y轴的刻度方向设置向内

labelsize=35;
tickedsize=35;
color1='#f0c808'
color2='#e07a5f'
color3='#3d405b'
color4='#81b29a'
import scipy.io as sio
from scipy.io import loadmat

```

```python
# Load data
File_name = 'synergy_result_LV1LOFA_to_LFFA_50ms_noresample.mat'
mean_monney_synergy = loadmat(File_name)['mean_monney_synergy']
mean_normal_synergy = loadmat(File_name)['mean_normal_synergy']
err_monney_synergy = loadmat(File_name)['err_monney_synergy']
err_normal_synergy = loadmat(File_name)['err_normal_synergy']
h_value = loadmat(File_name)['h_value_syn']
h_value1 = loadmat(File_name)['h_value_Monney_syn']
h_value2 = loadmat(File_name)['h_value_Normal_syn']
# Convert 2D arrays to 1D arrays
y1 = mean_monney_synergy.ravel()
yerr1 = err_monney_synergy.ravel()
y2 = mean_normal_synergy.ravel()
yerr2 = err_normal_synergy.ravel()
h0 = h_value.ravel()
h1 = h_value1.ravel()
h2 = h_value2.ravel()
# Extract the relevant time points
Time = np.arange(-0.95, 2.4999, (1/1200))
# Extract the relevant time points
#Time = np.arange(-0.50, 2.4999, 0.01)

# Create the plot
fig, axs = plt.subplots(figsize=(10, 7), dpi=300)

# Plot the first curve with shaded error area
axs.plot(1000*Time, y1, linewidth=3, color=color1, label='Mean Monney Synergy')
axs.fill_between(1000*Time, y1 - yerr1, y1 + yerr1, alpha=0.2, color=color1)

# Plot the second curve with shaded error area
axs.plot(1000*Time, y2, linewidth=3, color=color2, label='Mean Normal Synergy')
axs.fill_between(1000*Time, y2 - yerr2, y2 + yerr2, alpha=0.2, color=color2)


       
sig_y = 0.43  # Adjust the y-position of the line
for i, sig in enumerate(h1):
    if sig == 1:
        axs.scatter(1000*Time[i],y=sig_y, color=color1,s=5, marker='o')
sig_y = 0.435  # Adjust the y-position of the line
for i, sig in enumerate(h2):
    if sig == 1:
        axs.scatter(1000*Time[i],y=sig_y, color=color2,s=5, marker='o')

axs.spines['top'].set_visible(False)
axs.spines['right'].set_visible(False)
axs.spines['bottom'].set_linewidth(4)
axs.spines['left'].set_linewidth(4)
axs.tick_params(direction='in', width=2, length=4)
plt.xticks([0,500,1000,1500,2000],fontsize=20)
plt.yticks([0.1,0.2,0.3,0.4],fontsize=20)
plt.title(File_name)
plt.xlim([-100,2000])
plt.ylim([0.07,0.45])
```

```python
# add baseline
source='LV1LOFA'
target='LFFA'
File_name = 'synergy_result_'+ source +'_to_' +target+ '_50ms_noresample.mat'
File_name2='synergy_peak_'+ source +'_to_'+target +'_50ms_noresample.mat'
mean_monney_synergy = loadmat(File_name)['mean_monney_synergy']
mean_normal_synergy = loadmat(File_name)['mean_normal_synergy']
err_monney_synergy = loadmat(File_name)['err_monney_synergy']
err_normal_synergy = loadmat(File_name)['err_normal_synergy']

Monney_peak=loadmat(File_name2)['peakpointMonney']
Normal_peak=loadmat(File_name2)['peakpointNormal']
# Convert 2D arrays to 1D arrays
y1 = mean_monney_synergy.ravel()
yerr1 = err_monney_synergy.ravel()
y2 = mean_normal_synergy.ravel()
yerr2 = err_normal_synergy.ravel()
h0 = h_value.ravel()
h1 = h_value1.ravel()
h2 = h_value2.ravel()
# Extract the relevant time points
Time = np.arange(-0.95, 2.4999, (1/1200))
# Subtract baseline (e.g., mean of -0.5 to 0s)
baseline_window = (Time >=-0.084) & (Time <= 0)  # Define baseline time range
baseline1 = np.mean(y1[baseline_window])
baseline2 = np.mean(y2[baseline_window])

# Subtract the baseline from the synergy curves
y1 -= baseline1
y2 -= baseline2
# Create the plot
fig, axs = plt.subplots(figsize=(10, 7), dpi=300)

# Plot the first curve with shaded error area
axs.plot(1000*Time, y1, linewidth=3, color=color1, label='Mean Monney Synergy')
axs.fill_between(1000*Time, y1 - yerr1, y1 + yerr1, alpha=0.2, color=color1)

# Plot the second curve with shaded error area
axs.plot(1000*Time, y2, linewidth=3, color=color2, label='Mean Normal Synergy')
axs.fill_between(1000*Time, y2 - yerr2, y2 + yerr2, alpha=0.2, color=color2)



# Calculate mean and standard deviation
scatter_mean_monney = np.mean(Monney_peak)
scatter_std_monney = np.std(Monney_peak)
scatter_mean_normal = np.mean(Normal_peak)
scatter_std_normal = np.std(Normal_peak)

point_y1 = max(max(y1), max(y2)) + 0.020  # Adjust the y-position of the scatter points
point_y2 = max(max(y1), max(y2)) + 0.040  # Adjust the y-position of the scatter points

# Customize the plot
axs.spines['top'].set_visible(False)
axs.spines['right'].set_visible(False)
axs.spines['bottom'].set_linewidth(4)
axs.spines['left'].set_linewidth(4)
axs.tick_params(direction='in', width=2, length=4)
plt.xticks([0,100,200,300],fontsize=20)
plt.yticks([0,0.1,0.2,0.3],fontsize=20)
axs.tick_params(direction='in', width=2, length=4)
#plt.title(File_name)
plt.xlim([0,300])
# Show the plot
```
